## Average Score Per Term

```python
 
df['term_avg_score'] = df.groupby(['student_id', 'term'])['score'].transform('mean')
```

### Average score Per Subject

```python
df['subject_avg_score'] = df.groupby('subject')['score'].transform('mean')
```

### Overall Average Score per Student

```python
df['overall_avg_score'] = df.groupby('student_id')['score'].transform('mean')
```

### GPA (5-Point Scale)

```python
def score_to_gpa(score):
    if score >= 70:
        return 5.0
    elif score >= 60:
        return 4.0
    elif score >= 50:
        return 3.0
    elif score >= 45:
        return 2.0
    elif score >= 40:
        return 1.0
    else:
        return 0.0

df['gpa'] = df['score'].apply(score_to_gpa)
```

### GPA per Student per Term

```python
df['term_gpa'] = df.groupby(['student_id', 'term'])['gpa'].transform('mean')
```

### Overall GPA (CGPA)


```python
df['cgpa'] = df.groupby('student_id')['gpa'].transform('mean')
```

### Performance Category

```python
def performance_label(cgpa):
    if cgpa >= 4.5:
        return 'Excellent'
    elif cgpa >= 3.5:
        return 'Very Good'
    elif cgpa >= 2.5:
        return 'Good'
    elif cgpa >= 1.5:
        return 'Fair'
    else:
        return 'Poor'

df['performance_category'] = df['cgpa'].apply(performance_label)
```

### Visualization

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

   ### Subject line Performance Trendline 

```python
import matplotlib.pyplot as plt

subject_trend = df.groupby(['term', 'subject'])['score'].mean().reset_index()

plt.figure(figsize=(12,6))

for subject in subject_trend['subject'].unique():
    subset = subject_trend[subject_trend['subject'] == subject]
    plt.plot(subset['term'], subset['score'], marker='o', label=subject)

plt.xlabel("Term")
plt.ylabel("Average Score")
plt.title("Subject-Wise Performance Trend")
plt.legend()
plt.tight_layout()
plt.show()
```

### Attendance Heatmap

```python
import matplotlib.pyplot as plt
import numpy as np

pivot_att = df.pivot_table(
    index='student_id',
    columns='term',
    values='attendance_rate',
    aggfunc='mean'
)

plt.figure(figsize=(12, 6))
plt.imshow(pivot_att, aspect='auto')

plt.xticks(ticks=np.arange(len(pivot_att.columns)), labels=pivot_att.columns, rotation=45)
plt.yticks(ticks=np.arange(len(pivot_att.index)), labels=pivot_att.index)

plt.colorbar(label='Attendance %')
plt.title('Student Attendance Heatmap')
plt.xlabel('Term')
plt.ylabel('Student ID')
plt.tight_layout()
plt.show()
```
